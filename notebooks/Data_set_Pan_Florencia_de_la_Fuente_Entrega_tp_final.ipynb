# Actividad 5: Preguntas

## 1-¿Qué productos generan mayor Profit?

```python
pandata.groupby("Product")["Profit"].sum().sort_values(ascending=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculamos las ganancias por producto y seleccionamos el top 10 para el gráfico
ganancia_producto = pandata.groupby("Product")["Profit"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=ganancia_producto.values,
    y=ganancia_producto.index,
    hue=ganancia_producto.values,
    palette="Blues",
    legend=False
)

plt.title("Top 10 productos por ganancia total")
plt.xlabel("Ganancia total (Profit)")
plt.ylabel("Producto")
plt.tight_layout()
plt.show()
```

###Interpretación de los resultados:

El producto que genera mayor ganancia total es Coffee, con un Profit de 35.301,44, seguido por Cold Coffee, con 22.204,34, y Tea, con 14.168,81.

Algunos productos presentan valores negativos de Profit, como Black Forest Cake (-54.908,52), Chocolate Cake (-40.012,97) y Honey (-39.437,53).

Estos resultados permiten identificar los productos con mayor y menor Profit dentro del dataset. Sería interesante analizar qué factores explican los valores negativos, por ejemplo, los precios, los descuentos o los costos asociados a cada producto.

## 2-¿Las promociones influyen en la cantidad de Units_Sold?

```python
pandata.groupby("Promotion_Applied")["Units_Sold"].mean()
```

```python
pandata.groupby("Promotion_Applied")["Units_Sold"].sum()
```

```python
# Primero calculamos el promedio de unidades vendidas por promoción
ventas_promocion = pandata.groupby("Promotion_Applied")["Units_Sold"].mean()

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(7, 5))

sns.barplot(
    x=ventas_promocion.index,
    y=ventas_promocion.values,
    hue=ventas_promocion.index,
    palette={"No": "lightgray", "Yes": "teal"},
    legend=False
)

plt.title("Promedio de unidades vendidas según promoción")
plt.xlabel("¿Se aplicó una promoción?")
plt.ylabel("Promedio de unidades vendidas")
plt.tight_layout()
plt.show()
```

###Interpretación de los resultados:

Al comparar el promedio de Units_Sold entre los registros con promoción y sin promoción, se busca identificar si existen diferencias en la cantidad de unidades vendidas.

Si el promedio es mayor en los registros con promoción, esto indicaría una asociación entre las promociones y una mayor cantidad de unidades vendidas. Sin embargo, este resultado por sí solo no permite afirmar que las promociones sean la causa del aumento.

## 3-¿Qué días de la semana tienen mayor cantidad de ventas (Units_Sold)?

```python
pandata.groupby("Day_of_Week")["Units_Sold"].sum().sort_values(ascending=False)
```

```python
# Primero definimos la variable con la suma de unidades vendidas por día
ventas_dia = pandata.groupby("Day_of_Week")["Units_Sold"].sum().sort_values(ascending=False)

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.barplot(
    x=ventas_dia.index,
    y=ventas_dia.values,
    hue=ventas_dia.index,
    palette="Purples_r",
    legend=False
)

plt.title("Unidades vendidas por día de la semana")
plt.xlabel("Día de la semana")
plt.ylabel("Total de unidades vendidas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

###Interpretación de los resultados:

El sábado es el día con mayor cantidad de unidades vendidas, con un total de 24.895, seguido por el domingo, con 20.536 unidades.

En cambio, el lunes registra la menor cantidad de unidades vendidas, con 5.080.

Estos resultados muestran que las ventas se concentran principalmente durante el fin de semana. Esta información podría ser útil para analizar la planificación de la producción y la disponibilidad de stock según el día de la semana.
